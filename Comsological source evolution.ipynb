```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
def H(z: float) -> float:
    """Source evolution function from https://arxiv.org/pdf/astro-ph/0101216"""
    if z < 1.9:
        return (1 + z) ** 3
    elif z < 2.7:
        return (1 + 1.9) ** 3
    else:
        return (1 + 1.9) ** 3 * np.exp((2.7 - z) / 2.7)
```

```python
fig, ax = plt.subplots()

z_grid = np.logspace(-1.5, 1.5, 300)
ax.loglog(z_grid, [H(z) for z in z_grid])

ax.set_xlabel("$ z $")
ax.set_ylabel("$ \\mathcal{H} $")

plt.show()
```
